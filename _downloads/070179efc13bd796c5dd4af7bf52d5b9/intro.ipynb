```python
import torch
import numpy as np
#importing the libraries
```

```python
#Initializing a tensor

#We can initialize tensors in various way
# 1. From data
data = [[1,2], [3,4]]
x_data = torch.tensor(data) #seems like torch.sensor is a fx
```

```python
#we can create tensors w/ numpy
np_array = np.array(data)
x_np = torch.from_numpy(np_array)
```

```python
#Now what if you want data from another tensor?
x_ones = torch.ones_like(x_data) #this keeps whatever val of x
print(f"Ones Tensor: \n {x_ones} \n")

x_rand = torch.rand_like(x_data, dtype=torch.float) #overrides the data type
print(f"Randome Tensor: \n {x_rand} \n")
```

```python
#What about creating a tensor from random or constant values

#Lets introdue shape. Shape is a tuple of tensor dimensions. It basically
# determines the dimensionality of the output tensor.

shape = (2,3)
rand_tensor = torch.rand(shape)
ones_tensor = torch.ones(shape)
zeros_tensor = torch.zeros(shape)
twos_tensor = torch.full(shape, 2) #lets see what this does

#How this function works:
torch.ones((2,3)) #torch is the module, ones is the function call
#and 2,3 is the tuple describing what it is asking for
# this function is asking to create something with 2 rows and 3 cols


print(f"Random Tensor: \n {rand_tensor} \n")
print(f"Ones Tensor: \n {ones_tensor} \n")
print(f"Zeros Tensor: \n {zeros_tensor} \n")
print(f"Twos Tensor: \n {twos_tensor} \n")
```

```python
#Lets figure out the attributes of a tensor
tensor = torch.rand(3,4)

print(f"Shape of tensor: {tensor.shape}")
print(f"Datatype of tensor: {tensor.dtype}")
print(f"Device tensor is stored on: {tensor.device}")
```
